```python
import json
import matplotlib.pyplot as plt

# Load JSON data
with open("./Results_Final/New_Chunks_Evaluation_Results_Without_Finetuning.json", "r") as file:
    data = json.load(file)

# Initialize variables
scores = []
null_count = 0

# Process data
for entry in data:
    score = entry.get("score")
    if score is None:
        null_count += 1  # Track null scores
    else:
        scores.append(score)

# Calculate average score
average_score = sum(scores) / len(scores) if scores else 0

# Print results
print(f"Average Score: {average_score:.2f}")
print(f"Number of records with null scores: {null_count}")

# Plot Bar Graph of Score Distribution
plt.figure(figsize=(8, 5))
plt.hist(scores, bins=range(1, 7), edgecolor="black", alpha=0.7, rwidth=0.8)
plt.xticks(range(1, 6))
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.title("Score Distribution")
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Show the plot
plt.show()
```

# Finetuning > RAG

```python
import json
import matplotlib.pyplot as plt

def load_and_process_data(filepath):
    """Loads JSON data from a file and extracts scores."""
    with open(filepath, "r") as file:
        data = json.load(file)

    scores = []
    for entry in data:
        score = entry.get("score")
        if score is not None:
            scores.append(score)
    return scores

# Load data from both files
file1_scores = load_and_process_data("./Results_Final/New_Chunks_Evaluation_Results_With_Finetuning.json")
file2_scores = load_and_process_data("./Results_Final/New_Chunks_Evaluation_Results_Without_Finetuning.json")

# Create histograms
plt.figure(figsize=(12, 6))

plt.hist(file1_scores, bins=range(1, 7), alpha=0.6, label="With Finetuning", edgecolor='black', rwidth=0.8)
plt.hist(file2_scores, bins=range(1, 7), alpha=0.6, label="Without Finetuning", edgecolor='black', rwidth=0.8)

plt.xticks(range(1, 6))
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.title("Score Distribution Comparison")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
