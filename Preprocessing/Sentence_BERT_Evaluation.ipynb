# Sentence-BERT with finetuning

```python
import os
import torch
import json
import re
import csv
from transformers import AutoModelForCausalLM, AutoTokenizer
from dotenv import load_dotenv
from RAG_With_QWEN import RAG  # Import your RAG class
from RAG_With_QWEN import RailGuard
```

```python

# -------------------------------
# Load QA Dataset from CSV
# -------------------------------
def load_qa_dataset(csv_file):
    qa_dataset = []
    with open(csv_file, newline='', encoding="utf-8") as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            qa_dataset.append({
                "question": row["Question"],
                "expected_answer": row["Expected Answer"]
            })
    return qa_dataset

qa_dataset = load_qa_dataset("RAG_Evaluation_Dataset.csv")
```

```python
# -------------------------------
# Main Evaluation Routine
# -------------------------------
def main():
    rag_model = RAG()
    evaluations = []  
    railguard = RailGuard()
    
    for qa in qa_dataset:
        question = qa["question"]
        expected_answer = qa["expected_answer"]
        if railguard.railguard_eval(question):
            # Generate answer using your RAG model
            generated_answer = rag_model.generate_answer(question)

            #context 
            context = rag_model.get_context()

            # Print output in a structured format
            print("\n" + "="*80)
            print(f"**Question:** {question}")
            print(f"**Expected Answer:** {expected_answer}")
            print(f"**Generated Answer:** {generated_answer}")
            print(f"Context: {context}")
            print("="*80)

            # Store results for JSON output
            evaluations.append({
                "question": question,
                "expected_answer": expected_answer,
                "generated_answer": generated_answer,
                "context": context,
            })

    # Save evaluation results to a JSON file
    with open("./Results_Final/Generated_Answers_With_Context.json", "w") as outfile:
        json.dump(evaluations, outfile, indent=4)
    
    print("\nData for evaluation stored in./Results_Final/Generated_Answers_With_Context.json.")

if __name__ == "__main__":
    main()
```

```python
import json
from sentence_transformers import SentenceTransformer, util

# Load your QnA evaluation dataset from the JSON file
def load_results(json_file):
    with open(json_file, "r") as file:
        results = json.load(file)
    return results

# Save the results to a JSON file
def save_results(results, save_path):
    with open(save_path, "w") as file:
        json.dump(results, file, indent=4)
    print(f"Results saved to {save_path}")

# Initialize the Sentence-BERT model for semantic similarity
model = SentenceTransformer('all-MiniLM-L6-v2')

# Custom metric for "context relevance"
def context_relevance(generated_answer, context):
    # Compare the generated answer and context using cosine similarity
    context_embedding = model.encode(context, convert_to_tensor=True)
    answer_embedding = model.encode(generated_answer, convert_to_tensor=True)
    cosine_score = util.pytorch_cos_sim(answer_embedding, context_embedding)[0][0].item()
    return cosine_score

# Custom metric for "groundedness" (checks if the answer aligns with factual information or context)
def groundedness(generated_answer, context):
    # Check how grounded the answer is by comparing it with a reference context
    return context_relevance(generated_answer, context)

# Custom metric for "answer relevance" (how relevant the answer is to the question)
def answer_relevance(generated_answer, reference_answer):
    # Similarity between the answer and the expected answer (ground truth)
    return context_relevance(generated_answer, reference_answer)

# Custom metric for "faithfulness" (how true and consistent the answer is with the reference)
def faithfulness(generated_answer, reference_answer):
    # Faithfulness is measured by the cosine similarity between the generated and the reference answer
    return context_relevance(generated_answer, reference_answer)

# Load the dataset
results = load_results("./Results_Final/Generated_Answers_With_Context.json")

# Initialize a list to store evaluation results
final_results = []

# Loop through all questions in the dataset and calculate custom metrics
for result in results:
    generated_answer = result.get("generated_answer", "")
    reference_answer = result.get("expected_answer", "")
    context = result.get("context", "")  # Assuming contexts are stored in the dataset

    # Evaluate the custom metrics
    context_relevance_score = context_relevance(generated_answer, context)
    groundedness_score = groundedness(generated_answer, context)
    answer_relevance_score = answer_relevance(generated_answer, reference_answer)
    faithfulness_score = faithfulness(generated_answer, reference_answer)
    
    # Add the scores to the result dictionary
    result["context_relevance"] = context_relevance_score
    result["groundedness"] = groundedness_score
    result["answer_relevance"] = answer_relevance_score
    result["faithfulness"] = faithfulness_score
    
    # Append the result to the final results list
    final_results.append(result)

# Save the results to a JSON file
save_results(final_results, "./Results_Final/BERT-Sentence-scores.json")

# Calculate average scores for each metric
average_context_relevance = sum([r["context_relevance"] for r in final_results]) / len(final_results)
average_groundedness = sum([r["groundedness"] for r in final_results]) / len(final_results)
average_answer_relevance = sum([r["answer_relevance"] for r in final_results]) / len(final_results)
average_faithfulness = sum([r["faithfulness"] for r in final_results]) / len(final_results)

# Print the average scores
print(f"Average Context Relevance: {average_context_relevance}")
print(f"Average Groundedness: {average_groundedness}")
print(f"Average Answer Relevance: {average_answer_relevance}")
print(f"Average Faithfulness: {average_faithfulness}")
```

```python
scores = {}
scores["Average Context Relevance"] = average_context_relevance
scores["Average Groundedness"] = average_groundedness
scores["Average Answer Relevance"] = average_answer_relevance
scores["Average Faithfulness"] = average_faithfulness
save_results(scores, "./Results_Final/BERT-Sentence-average-scores.json")
```
